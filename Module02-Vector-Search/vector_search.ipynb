```python
from ingest import load_faq_data

documents = load_faq_data()
```

```python
documents[3]
```

```python
texts = []

for doc in documents:
    text = doc["question"] + " " + doc["answer"]
    texts.append(text)
```

```python
len(texts)
```

```python
from tqdm.auto import tqdm
```

```python
batch_size = 50
vectors = []

for i in tqdm(range(0, len(texts), batch_size)):
    batch = texts[i:i + batch_size]
    batch_vector = model.encode(batch)
    vectors.extend(batch_vector)

len(vectors)
```

```python
vectors[10].shape
```

```python
import numpy as np

X = np.array(vectors)
```

```python
X.shape
```

```python
# vector * vector multiplication

scores = []

for i in range(0, len(vectors)):
    score = vec1.dot(vectors[i])
    scores.append(score)

scores
```

```python
# another way

scores = [vec1.dot(X[i]) for i in range(len(vectors))]

scores
```

```python
# or use maxtrix * vector multiplication

scores = X.dot(vec1)
```

```python
scores
```

```python
index = np.argmax(scores)
```

```python
print("Index of highest cosine similarity: " + str(index))
print("Value of cosine similarity: " + str(scores[index]))
```

```python
documents[2]
```

```python
# to get top 5 not just first highest

top5 = np.argsort(scores)[-5:]
top5
```

```python
scores[top5]
    
```

```python
# reverse
top5 = top5[::-1]
top5
```

```python
for i in top5:
    print(scores[i])
    print(documents[i])
    print()
```

```python
top5 = np.argsort(-scores)[:5]
```

```python
for i in top5:
    print(scores[i])
    print(documents[i])
    print()
```

## Lesson 6

```python
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
openai_client = OpenAI()
```

```python
from ingest import load_faq_data, build_index

documents = load_faq_data()
index = build_index(documents)
```

```python
from rag_helper import RAGBase

assistant = RAGBase(
    index = index,
    llm_client = openai_client
)
```

```python
query = "I just discovered the course. Can I still join it?"
assistant.rag(query)
```

```python
class RAGVector(RAGBase):

    def __init__(self, embedder, **kwargs):
        super().__init__(**kwargs)
        self.embedder = embedder

    def search(self, query, num_results=5):
        query_vector = self.embedder.encode(query)
        filder_dict = {"course" : "llm-zoomcamp"}

        return self.index.search(
            query_vector,
            filter_dict=filder_dict,
            num_results=num_results
        )
```

```python
vector_assistant = RAGVector(
    embedder=model,
    index = vindex,
    llm_client = openai_client
)
```

```python
vector_assistant.rag("the program has already begun, can I still sign up?")
```
